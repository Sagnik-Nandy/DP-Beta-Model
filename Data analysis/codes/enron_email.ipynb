```python
import numpy as np
```

```python
def load_hypergraph(prefix):
    """Read a timestamped simplex dataset (prefix-nverts.txt, prefix-simplices.txt,
    prefix-times.txt) and return (hyperedges, times), where each hyperedge is a
    sorted tuple of node ids."""
    nverts = np.loadtxt(f"{prefix}-nverts.txt", dtype=int)
    simplices_raw = np.loadtxt(f"{prefix}-simplices.txt", dtype=int)
    times = np.loadtxt(f"{prefix}-times.txt", dtype=int)

    hyperedges = []
    current_pos = 0

    for count in nverts:
        nodes = simplices_raw[current_pos : current_pos + count]
        hyperedges.append(tuple(sorted(nodes)))
        current_pos += count

    return hyperedges, times
```
